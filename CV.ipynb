```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread(list(uploaded.keys())[0])

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

plt.imshow(gray, cmap="gray")
plt.axis("off")
plt.show()
```
